```python
#importando as bibliotecas
import pandas as pd
import scipy
from scipy.stats import variation
import numpy as np
```

```python
#Renomeando a planilha de dados
arquivo1='dados_projeto1.xlsx'
```

```python
#lendo os dados da planilha
df=pd.read_excel(arquivo1)
```

## Desvio Padrão

AMOSTRAL - DEVEMOS CALCULAR DESSA FORMA SE O CONJUNTO DE DADOS É UMA AMOSTRA

```python
#calcular o desvio padrão amostral da variável especificada
df.Idade.std()
```

POPULACIONAL - DEVEMOS CALCULAR DESSA FORMA SE O CONJUNTO DE DADOS É UMA POPULAÇÃO

```python
#calcular o desvio padrão populacional da variável especificada
df.Idade.std(ddof=0)
```

```python
#Quando o data set possui mais de uma variável númerica, é possível calcular
# o desvio padrão de todas as variáveis simultaneamente. No arquivo só pode ter v. quantitativas
arquivo2='exemplo2 tabela.xlsx'
df2=pd.read_excel(arquivo2)
df2.std()
```

## Coeficiente de variação

AMOSTRAL - DEVEMOS CALCULAR DESSA FORMA SE O CONJUNTO DE DADOS É UMA AMOSTRA

```python
#coeficiente de variação (amostral)
scipy.stats.variation(df.Idade,ddof=1)*100
```

POPULACIONAL - DEVEMOS CALCULAR DESSA FORMA SE O CONJUNTO DE DADOS É UMA POPULAÇÃO

```python
#coeficiente de variação (populacional)
scipy.stats.variation(df.Idade,ddof=0)*100
```
